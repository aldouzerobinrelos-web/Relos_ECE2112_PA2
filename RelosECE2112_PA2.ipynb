```python
import numpy as np
```

# **A. REPRODUCIBLE NORMALIZATION**



```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
x = X.mean()
σ = X.std()
X_normalized = (X - x)/σ
x_normalized = X_normalized.mean()
σ_normalized = X_normalized.std()
```

**X**




```python
print(X)
```

**X_normalized**

```python
print(X_normalized)
```

**x_normalized**

```python
print(x_normalized)
```

**σ_normalized**

```python
print(σ_normalized)
```

```python
np.save("X_normalized.npy",X_normalized)
```

# **B. CUBES DIVISIBLE BY 4 PROBLEM**

```python
c = (np.arange(1, 101)**3).reshape(10, 10)
div4 = c%4 == 0
div_by_4 = c[div4].reshape(5,10)
```

**Shape of C**

```python
print(c.shape)
```

**div_by_4 array**

```python
print(div_by_4)
```

**Number of Elements**

```python
display(div_by_4.size)
```

```python
np.save("div_by_4.npy",div_by_4)
```

# **C. ABOVE-MEAN SQUARES PROBLEM**

```python
S = (np.arange(1, 37)**2).reshape(6,6)
S_mean = np.mean(S)
above_mean = S[S > S_mean].reshape(3,5)
```

**Array of S**

```python
display(S)
```

**Mean of S**

```python
display(S_mean)
```

**above_mean array**

```python
display(above_mean)
```

```python
np.save("above_mean.npy",above_mean)
```
